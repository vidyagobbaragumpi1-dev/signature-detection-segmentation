```python
import os
import cv2
import numpy as np
from ultralytics import YOLO

# -----------------------------
# Load trained detection model
# -----------------------------
model = YOLO("D:/runs/detect/Train2/weights/best.pt")

# -----------------------------
# Dataset paths
# -----------------------------
image_folder = r"D:/signature/images/Test"
label_folder = r"D:/signature/labels/Test"

# -----------------------------
# Convert YOLO label to bbox
# -----------------------------
def yolo_to_bbox(label, W, H):
    cls, x, y, w, h = map(float, label.split())

    x1 = int((x - w/2) * W)
    y1 = int((y - h/2) * H)
    x2 = int((x + w/2) * W)
    y2 = int((y + h/2) * H)

    return [x1, y1, x2, y2]

# -----------------------------
# IoU Function
# -----------------------------
def compute_iou(boxA, boxB):

    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])

    inter = max(0, xB-xA) * max(0, yB-yA)

    areaA = (boxA[2]-boxA[0]) * (boxA[3]-boxA[1])
    areaB = (boxB[2]-boxB[0]) * (boxB[3]-boxB[1])

    union = areaA + areaB - inter

    if union == 0:
        return 0

    return inter / union


# -----------------------------
# Store IoUs
# -----------------------------
all_ious = []

# -----------------------------
# Process each image
# -----------------------------
for file in os.listdir(image_folder):

    if not file.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    img_path = os.path.join(image_folder, file)

    image = cv2.imread(img_path)

    if image is None:
        continue

    H, W = image.shape[:2]

    # Ground truth label
    label_path = os.path.join(
        label_folder,
        os.path.splitext(file)[0] + ".txt"
    )

    if not os.path.exists(label_path):
        print(f"Missing label: {file}")
        continue

    with open(label_path) as f:
        gt_boxes = [yolo_to_bbox(line, W, H) for line in f.readlines()]

    # Prediction
    results = model.predict(
        source=img_path,
        conf=0.25,
        verbose=False
    )

    pred_boxes = results[0].boxes.xyxy.cpu().numpy()

    if len(pred_boxes) == 0:
        print(f"{file} : No detections")
        continue

    # Match each prediction with best GT
    for i, pred in enumerate(pred_boxes):

        best_iou = 0

        for gt in gt_boxes:

            iou = compute_iou(pred, gt)

            if iou > best_iou:
                best_iou = iou

        all_ious.append(best_iou)

        print(f"{file} -> Detection {i+1}: IoU = {best_iou:.4f}")

# -----------------------------
# Overall Results
# -----------------------------
print("\n========== OVERALL RESULTS ==========")

if len(all_ious):

    print(f"Mean IoU : {np.mean(all_ious):.4f}")
    print(f"Max IoU  : {np.max(all_ious):.4f}")
    print(f"Min IoU  : {np.min(all_ious):.4f}")
    print(f"Total Detections : {len(all_ious)}")

else:
    print("No detections found.")
```

```python
import os
import cv2
from ultralytics import YOLO

# Load trained model
model = YOLO("D:/runs/detect/Train2/weights/best.pt")

# Input images
image_folder = "D:/signature/images/train"                        #cropped train images

# Output folder
output_folder = "D:/segmentation/images/train"
os.makedirs(output_folder, exist_ok=True)

# Process all images
for file in os.listdir(image_folder):

    if not file.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    img_path = os.path.join(image_folder, file)

    # Read image
    image = cv2.imread(img_path)

    # Run detection
    results = model(img_path)

    if len(results[0].boxes) == 0:
        print(f"No signature detected: {file}")
        continue

    # Crop each detected signature
    for i, box in enumerate(results[0].boxes.xyxy.cpu().numpy()):

        x1, y1, x2, y2 = map(int, box)

        # Crop
        crop = image[y1:y2, x1:x2]

        # Save
        save_name = f"{file}_sig_{i}.png"
        save_path = os.path.join(output_folder, save_name)

        cv2.imwrite(save_path, crop)

        print(f"Saved: {save_name}")

print("✅ Signature extraction completed")
```

```python
import os
import json

# Folder containing Labelme JSON files
json_folder = "D:/segmentation/labels/train"

# Output folder for YOLO TXT labels
txt_folder = "D:/output/train"

os.makedirs(txt_folder, exist_ok=True)

# Class mapping
class_map = {
    "signature": 0
}

for file in os.listdir(json_folder):

    if not file.endswith(".json"):
        continue

    json_path = os.path.join(json_folder, file)

    with open(json_path, "r") as f:
        data = json.load(f)

    img_w = data["imageWidth"]
    img_h = data["imageHeight"]

    txt_name = os.path.splitext(file)[0] + ".txt"
    txt_path = os.path.join(txt_folder, txt_name)

    with open(txt_path, "w") as out:

        for shape in data["shapes"]:

            label = shape["label"]

            if label not in class_map:
                continue

            class_id = class_map[label]

            points = shape["points"]

            coords = []

            for x, y in points:

                x = x / img_w
                y = y / img_h

                coords.append(f"{x:.6f}")
                coords.append(f"{y:.6f}")

            out.write(f"{class_id} " + " ".join(coords) + "\n")

print("✅ Conversion completed successfully!")
print("TXT files saved in:", txt_folder)
```

```python
import os
import cv2
import numpy as np
from ultralytics import YOLO

# -----------------------------
# Load trained segmentation model
# -----------------------------
model = YOLO("D:/runs/segment/train6/weights/best.pt")

# -----------------------------
# Input and Output folders
# -----------------------------
input_folder = r"D:/segmentation/images/test"
output_folder = r"D:/segmentation/extracted"

os.makedirs(output_folder, exist_ok=True)

# -----------------------------
# Process each image
# -----------------------------
for file in os.listdir(input_folder):

    if not file.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    img_path = os.path.join(input_folder, file)

    image = cv2.imread(img_path)

    if image is None:
        continue

    h, w = image.shape[:2]

    # -----------------------------
    # Predict segmentation
    # -----------------------------
    results = model.predict(img_path, conf=0.25, verbose=False)

    if results[0].masks is None:
        print(f"No signature detected : {file}")
        continue

    # Empty mask
    mask = np.zeros((h, w), dtype=np.uint8)

    # Draw predicted polygons
    for poly in results[0].masks.xy:

        poly = np.array(poly, dtype=np.int32)

        cv2.fillPoly(mask, [poly], 255)

    # -----------------------------
    # Extract signature
    # -----------------------------
    extracted = cv2.bitwise_and(image, image, mask=mask)

    # Optional: white background instead of black
    white = np.full_like(image, 255)
    white_bg = np.where(mask[:, :, None] == 255, extracted, white)

    # -----------------------------
    # Save outputs
    # -----------------------------
    cv2.imwrite(os.path.join(output_folder, file), white_bg)

    print(f"Extracted: {file}")

print("\nExtraction Completed Successfully!")
```

```python
import cv2
import os

input_folder = r"D:/segmentation/extracted"
output_folder = r"D:/segmentation/extracted1"

os.makedirs(output_folder, exist_ok=True)

for img_name in os.listdir(input_folder):

    if not img_name.lower().endswith((".png", ".jpg", ".jpeg")):
        continue

    img_path = os.path.join(input_folder, img_name)

    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

    # Invert image
    white_signature = cv2.bitwise_not(img)

    save_path = os.path.join(output_folder, img_name)
    cv2.imwrite(save_path, white_signature)

print("All images converted successfully.")
```
